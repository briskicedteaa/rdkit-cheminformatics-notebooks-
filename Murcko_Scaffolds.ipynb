```python
df = pd.read_csv('/content/File.csv', sep =';' )
df = df.drop(columns=df.columns.difference(['Target ChEMBL ID', 'Smiles']))
df = df.dropna()
df
```

```python
for index, row in df.iterrows():
  target_ID = row['Target ChEMBL ID']
  smiles_string = row['Smiles']
  mol = Chem.MolFromSmiles(smiles_string)


valid_mask = [Chem.MolFromSmiles(s) is not None for s in df['Smiles']]

df = df[valid_mask]

df.to_csv('File.csv', index=False)

display(df)
```

```python
from rdkit.Chem.Scaffolds import MurckoScaffold

csv_data = pd.read_csv("File.csv")

def extract_single_scaffold(smiles_text):
    mol = Chem.MolFromSmiles(smiles_text)
    if mol is not None:
        scaffold_obj = MurckoScaffold.GetScaffoldForMol(mol)
        return Chem.MolToSmiles(scaffold_obj)
    return None

csv_data['scaffold_results'] = csv_data['Smiles'].apply(extract_single_scaffold)

df['Murcko Scaffolds'] = df['Smiles'].apply(extract_single_scaffold)

display(df)
```
