```python
ligandlist = [
    "Cn1c(=O)c2c(ncn2C)n(C)c1=O",

    "CC(=O)Oc1ccccc1C(=O)O",

    "CC(=O)NC1=CC=C(C=C1)O",

    "CC(C)CC1=CC=C(C=C1)C(C)C(=O)O",

    "CN1CCC[C@H]1c2cccnc2",

    "NCCc1ccc(O)c(O)c1",

    "NCCc1c[nH]c2ccc(O)cc12",

    "C1CC"

]

```

```python
for i, smiles_string in enumerate(ligandlist):
  mol = Chem.MolFromSmiles(smiles_string)
  if mol:
    numheavyatoms = mol.GetNumHeavyAtoms()
    numrings = mol.GetRingInfo().NumRings()
    print(f"{i+1}, {smiles_string}, {numheavyatoms}, {numrings}")

  else: print(f"{i+1}, {smiles_string}): Invalid string")
```

```python
from rdkit.Chem import PandasTools

df = pd.read_csv("Ligands1 - Sheet1 2.csv")

df['Formula'] = df['SMILES'].apply(lambda x: Chem.rdMolDescriptors.CalcMolFormula(Chem.MolFromSmiles(x)) if pd.notna(x) else None)

mols = df['SMILES'].apply(lambda x: Chem.MolFromSmiles(x) if pd.notna(x) else None)

df['Formula'] = mols.apply(lambda m: Chem.rdMolDescriptors.CalcMolFormula(m) if m else None)
df['Has_Ring'] = mols.apply(lambda m: m.GetRingInfo().NumRings() > 0 if m else False)
df['Is_Aromatic'] = mols.apply(lambda m: any(a.GetIsAromatic() for a in m.GetAtoms()) if m else False)
df['Canonical_SMILES'] = mols.apply(lambda m: Chem.MolToSmiles(m, canonical=True) if m else None)

df
```
