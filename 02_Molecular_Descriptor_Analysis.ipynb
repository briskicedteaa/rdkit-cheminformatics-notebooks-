```python
from rdkit import Chem
from rdkit.Chem import Descriptors

df['NumHeavyAtoms'] = None
df['LogP'] = None
df['HBA'] = None
df['HBD'] = None
df['TPSA'] = None
df['RotatableBonds'] = None
df['NumRings'] = None
df['MolWt'] = None
df['Parsing Succesful?'] = True

for index, row in df.iterrows():
  compound_name = row['Compound']
  smiles_string = row['SMILES']

  mol = Chem.MolFromSmiles(smiles_string)
  if mol:
    numheavyatoms = mol.GetNumHeavyAtoms()
    logp = Descriptors.MolLogP(mol)
    hba = Descriptors.NumHAcceptors(mol)
    hbd = Descriptors.NumHDonors(mol)
    tpsa = Descriptors.TPSA(mol)
    MolWt = Descriptors.MolWt(mol)
    rot_bonds = Descriptors.NumRotatableBonds(mol)
    numrings = mol.GetRingInfo().NumRings()

    df.loc[index, 'NumHeavyAtoms'] = numheavyatoms
    df.loc[index, 'LogP'] = logp
    df.loc[index, 'HBA'] = hba
    df.loc[index, 'HBD'] = hbd
    df.loc[index, 'TPSA'] = tpsa
    df.loc[index, 'RotatableBonds'] = rot_bonds
    df.loc[index, 'NumRings'] = numrings
    df.loc[index, 'MolWt'] = MolWt

  else:
    print(f"Compound {compound_name} (SMILES: {smiles_string}): Invalid SMILES string")
    df.loc[index, 'Parsing Succesful?'] = False

display(df)
```

```python
import pandas as pd

descriptors = ['NumHeavyAtoms', 'MolWt', 'LogP', 'HBA', 'HBD', 'TPSA', 'RotatableBonds', 'NumRings']
summary_data = []

for descriptor in descriptors:
    df_filtered = df[df[descriptor].notna()]
    if not df_filtered.empty:
        min_value_row = df_filtered.loc[df_filtered[descriptor].idxmin()]
        max_value_row = df_filtered.loc[df_filtered[descriptor].idxmax()]

        summary_data.append({
            'Descriptor': descriptor,
            'Min Value Compound': min_value_row['Compound'],
            'Min Value': min_value_row[descriptor],
            'Max Value Compound': max_value_row['Compound'],
            'Max Value': max_value_row[descriptor]
        })

summary_df = pd.DataFrame(summary_data)
display(summary_df)
```

```python
from rdkit import Chem
from rdkit.Chem import Descriptors

df['LogP'] = None
df['TPSA'] = None
df['RotatableBonds'] = None
df['MolWt'] = None

for index, row in df.iterrows():
  compound_name = row['Compound']
  smiles_string = row['SMILES']

  mol = Chem.MolFromSmiles(smiles_string)
  if mol:
    logp = Descriptors.MolLogP(mol)
    tpsa = Descriptors.TPSA(mol)
    MolWt = Descriptors.MolWt(mol)
    rot_bonds = Descriptors.NumRotatableBonds(mol)

    df.loc[index, 'LogP'] = logp
    df.loc[index, 'TPSA'] = tpsa
    df.loc[index, 'RotatableBonds'] = rot_bonds
    df.loc[index, 'MolWt'] = MolWt

  else:
    print(f"Compound {compound_name} (SMILES: {smiles_string}): Invalid SMILES string")
    df.loc[index, 'Parsing Succesful?'] = False

df.drop(columns = ['HBA', 'HBD', 'NumHeavyAtoms', 'NumRings', 'Parsing Succesful?'], inplace = True)

display(df)
```
