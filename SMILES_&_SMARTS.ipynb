```python
df = pd.read_csv('/content/Ligand.csv')

df
```

```python
from rdkit.Chem import PandasTools

for index, row in df.iterrows():
  compound_name = row['Compound']
  smiles_string = row['SMILES']
  mol = Chem.MolFromSmiles(smiles_string)
  PandasTools.AddMoleculeColumnToFrame(df, smilesCol='SMILES', molCol='ROMol')
  PandasTools.WriteSDF(df, 'Ligand.sdf', molColName='ROMol', idName='Compound', properties=list(df.columns))

display(df)
```

```python
suppl = Chem.SDMolSupplier('Ligand.sdf')

patt = Chem.MolFromSmarts('c1ccccc1')
matches = []
for mol in suppl:
  if mol.HasSubstructMatch(patt):
    matches.append(mol)

len(matches)
display(df)
```
