```python
!pip install rdkit
from rdkit import Chem
import pandas as pd
from rdkit import RDLogger
```

```python
df = pd.read_csv('/content/File.csv', sep =';' )
df = df.drop(columns=df.columns.difference(['Target ChEMBL ID', 'Smiles']))
df = df.dropna()
df
```

```python
for index, row in df.iterrows():
  target_ID = row['Target ChEMBL ID']
  smiles_string = row['Smiles']
  mol = Chem.MolFromSmiles(smiles_string)

  RDLogger.DisableLog('rdkit.Chem.Outputs')

valid_mask = [Chem.MolFromSmiles(s) is not None for s in df['Smiles']]

df = df[valid_mask]

df.to_csv('File.csv', index=False)

display(df)
```

```python
from rdkit.Chem import rdFingerprintGenerator

mfpgen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=256)

mol = Chem.MolFromSmiles(smiles_string)

fp = mfpgen.GetFingerprint(mol)
fp.ToList()[:10]
df['Fingerprints'] = fp
display(df)
```

```python
from rdkit import DataStructs

query_mol = Chem.MolFromSmiles("CN1CCI[C@@]2(C)C3Cc(OC(=O)NCCCCCCCCCN4CCOCC4)cCC3N(C)[C@@H]12")
mfpgen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=256)
query_fp = mfpgen.GetFingerprint(query_mol)

similarities = []
for index, row_data in df.iterrows():
    mol = Chem.MolFromSmiles(row_data['Smiles'])
    if mol is not None:
        fp = mfpgen.GetFingerprint(mol)
        similarity = DataStructs.TanimotoSimilarity(query_fp, fp)
        similarities.append(similarity)
    else:
        similarities.append(None)

df['similarity'] = similarities

display(df)
```
