```python
df = pd.read_csv('/content/File.csv', sep =';' )
df = df.drop(columns=df.columns.difference(['Target ChEMBL ID', 'Smiles']))
df = df.dropna()
df
```

```python
from rdkit.Chem import Descriptors

df['NumHeavyAtoms'] = None
df['LogP'] = None
df['HBA'] = None
df['HBD'] = None
df['TPSA'] = None
df['RotatableBonds'] = None
df['NumRings'] = None
df['MolWt'] = None
df['Parsing Succesful?'] = True

for index, row in df.iterrows():
  compound_name = row['Target ChEMBL ID']
  smiles_string = row['Smiles']

  mol = Chem.MolFromSmiles(smiles_string)
  if mol:
    numheavyatoms = mol.GetNumHeavyAtoms()
    logp = Descriptors.MolLogP(mol)
    hba = Descriptors.NumHAcceptors(mol)
    hbd = Descriptors.NumHDonors(mol)
    tpsa = Descriptors.TPSA(mol)
    MolWt = Descriptors.MolWt(mol)
    rot_bonds = Descriptors.NumRotatableBonds(mol)
    numrings = mol.GetRingInfo().NumRings()

    df.loc[index, 'NumHeavyAtoms'] = numheavyatoms
    df.loc[index, 'LogP'] = logp
    df.loc[index, 'HBA'] = hba
    df.loc[index, 'HBD'] = hbd
    df.loc[index, 'TPSA'] = tpsa
    df.loc[index, 'RotatableBonds'] = rot_bonds
    df.loc[index, 'NumRings'] = numrings
    df.loc[index, 'MolWt'] = MolWt

  else:
    print(f"Compound {compound_name} (Smiles: {smiles_string}): Invalid Smiles string")
    df.loc[index, 'Parsing Succesful?'] = False

display(df)
```
