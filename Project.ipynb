```python
df = pd.read_csv('/content/File.csv', sep =';' )
df = df.drop(columns=df.columns.difference(['Target ChEMBL ID', 'Smiles']))
df = df.dropna()
df
```

```python
for index, row in df.iterrows():
  target_ID = row['Target ChEMBL ID']
  smiles_string = row['Smiles']
  mol = Chem.MolFromSmiles(smiles_string)

valid_mask = [Chem.MolFromSmiles(s) is not None for s in df['Smiles']]

df = df[valid_mask]

df.to_csv('File.csv', index=False)

display(df)
```

```python
from rdkit.Chem import Descriptors

df['NumHeavyAtoms'] = None
df['LogP'] = None
df['HBA'] = None
df['HBD'] = None
df['TPSA'] = None
df['RotatableBonds'] = None
df['NumRings'] = None
df['MolWt'] = None
df['Parsing Succesful?'] = True

for index, row in df.iterrows():
  compound_name = row['Target ChEMBL ID']
  smiles_string = row['Smiles']

  mol = Chem.MolFromSmiles(smiles_string)
  if mol:
    numheavyatoms = mol.GetNumHeavyAtoms()
    logp = Descriptors.MolLogP(mol)
    hba = Descriptors.NumHAcceptors(mol)
    hbd = Descriptors.NumHDonors(mol)
    tpsa = Descriptors.TPSA(mol)
    MolWt = Descriptors.MolWt(mol)
    rot_bonds = Descriptors.NumRotatableBonds(mol)
    numrings = mol.GetRingInfo().NumRings()

    df.loc[index, 'NumHeavyAtoms'] = numheavyatoms
    df.loc[index, 'LogP'] = logp
    df.loc[index, 'HBA'] = hba
    df.loc[index, 'HBD'] = hbd
    df.loc[index, 'TPSA'] = tpsa
    df.loc[index, 'RotatableBonds'] = rot_bonds
    df.loc[index, 'NumRings'] = numrings
    df.loc[index, 'MolWt'] = MolWt

  else:
    print(f"Compound {compound_name} (Smiles: {smiles_string}): Invalid Smiles string")
    df.loc[index, 'Parsing Succesful?'] = False

display(df)
```

```python
from rdkit.Chem import rdFingerprintGenerator

mfpgen = rdFingerprintGenerator.GetMorganGenerator(radius=2, fpSize=256)

mol = Chem.MolFromSmiles(smiles_string)

fp = mfpgen.GetFingerprint(mol)
fp.ToList()[:10]
df['Fingerprints'] = fp
display(df)
```

```python
from rdkit.Chem.Scaffolds import MurckoScaffold

csv_data = pd.read_csv("File.csv")

def extract_single_scaffold(smiles_text):
    mol = Chem.MolFromSmiles(smiles_text)
    if mol is not None:
        scaffold_obj = MurckoScaffold.GetScaffoldForMol(mol)
        return Chem.MolToSmiles(scaffold_obj)
    return None

csv_data['scaffold_results'] = csv_data['Smiles'].apply(extract_single_scaffold)

df['Murcko Scaffolds'] = df['Smiles'].apply(extract_single_scaffold)

display(df)
```

```python
counts = df['Murcko Scaffolds'].value_counts()
top_scaffold = counts.idxmax()
top_10 = df['Murcko Scaffolds'].value_counts().head(10)

display(counts)
print(top_scaffold)
display(top_10)
```
